```python
def dfs(jug1, jug2, visited, path):
    if (jug1, jug2) in visited:
        return False
    
    visited.add((jug1, jug2))
    path.append((jug1, jug2))
    
    if jug1 == 2 or jug2 == 2:
        return True
    
    max_jug1 = 4
    max_jug2 = 3
    
    next_states = [
        (max_jug1, jug2),  
        (jug1, max_jug2),  
        (0, jug2),         
        (jug1, 0),         
        (jug1 - min(jug1, max_jug2 - jug2), jug2 + min(jug1, max_jug2 - jug2)),  
        (jug1 + min(jug2, max_jug1 - jug1), jug2 - min(jug2, max_jug1 - jug1))   
    ]
    
    rules = [
        "Fill Jug1",
        "Fill Jug2",
        "Empty Jug1",
        "Empty Jug2",
        "Pour Jug1 -> Jug2",
        "Pour Jug2 -> Jug1"
    ]
    
    for i in range(len(next_states)):
        next_jug1, next_jug2 = next_states[i]
        if dfs(next_jug1, next_jug2, visited, path):
            print("Rule Applied:", rules[i])
            print("State:", next_jug1, next_jug2)
            return True
    
    path.pop()
    return False


visited = set()
path = []

dfs(0, 0, visited, path)

print("\nSolution Path:")
for state in path:
    print(state)
```
